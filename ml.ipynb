```python
a = np.array([1,2,3])
b = np.array([6,7,3])

sum = np.add(a,b)
print(sum)
sub = np.subtract(a,b)
print(sub)
mul = np.multiply(a,b)
print(mul)
div = np.divide(a,b)
print(div)

l = np.power(a,b)
print(l)

m = np.mod(a,b)
print(m)
```
