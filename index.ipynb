```python
# multilevel inheritance;;

class GrandA:
    def grd(self):
        return "grandFather is calling..."
class Parent(GrandA):
    pass

class Child(Parent):
    def show(self):
        return "hell child how are you?"

h1 = Child()
print(h1.grd() )
```

```python
# abstraction

class Employee: # abstract class
    def __init__(self,name,age):
        self.name = name 
        self.age = age 
    def show(self): # abstract method
        pass

class SubEmployee(Employee):
    def show(self):
        return self.name
s1 = SubEmployee("asif", 29)
s1.show()
```

```python
# recursion

def factorial(n):
    if(n == 0 or n == 1):
        return 1 
    else:
        return n * factorial(n-1)

print(factorial(6))
```
